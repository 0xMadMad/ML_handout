# split data

```python
features_df = df[["ENGINESIZE", "CYLINDERS", "FUELCONSUMPTION_CITY", "FUELCONSUMPTION_HWY", "FUELCONSUMPTION_COMB", "FUELCONSUMPTION_COMB_MPG", "CO2EMISSIONS"]]
features_df.head()
```

```python
features_df.info()
```

### simplest way to split data for small datasets is masking

each run creates different mask list => different train/test dataset

```python

mask = np.random.rand(len(features_df)) < 0.8  # creating a list of True and False values with same length as the dataset. "<0.8"=="80%" are true
mask
```

```python
train = features_df[mask] 
test = features_df[~mask]
```

```python
train
```

```python
test
```

# train with `ENGINESIZE`

```python
from sklearn import linear_model
```

```python
e_train_x = np.asanyarray(train[['ENGINESIZE']])  # "asanyarray" np array to general array
e_train_y = np.asanyarray(train[['CO2EMISSIONS']])

e_reg = linear_model.LinearRegression()
e_reg.fit (e_train_x, e_train_y)

print ('coefficients (slope/theta1): ', e_reg.coef_[0][0])
print ('Intercept (theta0): ',e_reg.intercept_[0])
```

```python
plt.scatter(train.ENGINESIZE, train.CO2EMISSIONS,  color='blue')
plt.plot(e_train_x, (e_reg.coef_[0][0]*e_train_x)+e_reg.intercept_[0], 'r-')
plt.xlabel("ENGINESIZE")
plt.ylabel("EMISSION")
```

## test

```python
from sklearn.metrics import r2_score

e_test_x = np.asanyarray(test[['ENGINESIZE']])
e_test_y = np.asanyarray(test[['CO2EMISSIONS']])
e_test_y_ = e_reg.predict(e_test_x)

print("score : %.2f" % e_reg.score(e_test_x, e_test_y))
print("R2-score: %.2f" % r2_score(e_test_y , e_test_y_) )
```
