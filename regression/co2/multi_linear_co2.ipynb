# split data

### simplest way to split data for small datasets is masking

each run creates different mask list => different train/test dataset

```python
features_df = df[["ENGINESIZE", "FUELCONSUMPTION_CITY", "FUELCONSUMPTION_HWY", "FUELCONSUMPTION_COMB", "CO2EMISSIONS"]]
features_df.head()
```

```python

mask = np.random.rand(len(features_df)) < 0.8  # creating a list of True and False values with same length as the dataset. "<0.8"=="80%" are true
mask
```

```python
train = features_df[mask] 
test = features_df[~mask]
```

```python
from sklearn.metrics import r2_score

chc_test_x = np.asanyarray(test[["FUELCONSUMPTION_CITY", "FUELCONSUMPTION_HWY", "FUELCONSUMPTION_COMB"]])
chc_test_y = np.asanyarray(test[["CO2EMISSIONS"]])
chc_test_y_ = chc_reg.predict(chc_test_x)  # it's y hat

print("score: %.2f" % chc_reg.score(chc_test_x, chc_test_y))
print("R2-score: %.2f" % r2_score(chc_test_y , chc_test_y_) )
```
